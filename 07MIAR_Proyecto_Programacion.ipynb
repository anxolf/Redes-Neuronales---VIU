## 1.2 Carga del conjunto de datos

La descarga se hace desde el propio notebook mediante una función reutilizable y **sin necesidad de credenciales** (no hay que subir `kaggle.json`):

2. **Fuente de respaldo**: si la descarga desde Kaggle falla (sin conexión a Kaggle, límite de peticiones, etc.), se descarga automáticamente el conjunto *flower_photos* publicado por TensorFlow, que contiene las mismas cinco especies de flores. La celda indica qué fuente se ha usado.

En ambos casos el notebook se ejecuta de principio a fin sin intervención manual, tanto en Google Colab como en un Jupyter local.

```python
# Descarga SIN credenciales: kagglehub permite bajar datasets publicos de Kaggle de forma anonima.
# Si kagglehub no esta instalado (p. ej. en un Jupyter local), se instala de forma silenciosa.
import sys
try:
    import kagglehub
except ImportError:
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'kagglehub'], check=False)
    try:
        import kagglehub
    except ImportError:
        kagglehub = None
print('kagglehub disponible:', kagglehub is not None)
```

```python
IMG_EXT = {'.jpg', '.jpeg', '.png'}

def download_dataset(kaggle_slug='alxmamaev/flowers-recognition'):
    # Devuelve la carpeta con el dataset descargado. No requiere credenciales:
    #   1) Kaggle (descarga anonima de un dataset publico con kagglehub)
    #   2) Respaldo: flower_photos de TensorFlow (mismas 5 especies), si Kaggle no responde
    global DATA_SOURCE
    if kagglehub is not None:
        try:
            path = Path(kagglehub.dataset_download(kaggle_slug))
            DATA_SOURCE = 'Kaggle: ' + kaggle_slug
            print('Dataset descargado desde Kaggle en', path)
            return path
        except Exception as e:
            print('No se ha podido descargar desde Kaggle (%s). Se usa la fuente de respaldo.' % type(e).__name__)
    archive = keras.utils.get_file(
        'flower_photos',
        origin='https://storage.googleapis.com/download.tensorflow.org/example_images/flower_photos.tgz',
        untar=True, cache_dir=str(Path.cwd()), cache_subdir='datasets')
    DATA_SOURCE = 'TensorFlow: flower_photos'
    print('Dataset descargado desde la fuente de respaldo en', archive)
    return Path(archive).parent

def find_class_root(root):
    # Localiza la carpeta cuyas subcarpetas contienen directamente las imagenes (una subcarpeta por clase).
    best, best_n = None, 0
    for d, subdirs, _ in os.walk(root):
        n = 0
        for s in subdirs:
            files_in = os.listdir(os.path.join(d, s))
            if any(Path(f).suffix.lower() in IMG_EXT for f in files_in):
                n += 1
        if n > best_n:
            best, best_n = d, n
    assert best is not None and best_n >= 2, 'No se ha encontrado una estructura carpeta-por-clase'
    return Path(best)

DATA_DIR = download_dataset('alxmamaev/flowers-recognition')
print('Fuente de los datos:', DATA_SOURCE)
CLASS_ROOT = find_class_root(DATA_DIR)
print('Carpeta con las clases:', CLASS_ROOT)
print('Subcarpetas:', sorted(p.name for p in CLASS_ROOT.iterdir() if p.is_dir()))
```

## 1.3 Inspección y análisis del conjunto de datos

Se construye un índice con una fila por imagen: ruta, clase, ancho, alto, modo de color y *hash* MD5 del fichero. Abrir cada imagen permite detectar ficheros corruptos y el *hash* permite detectar **duplicados exactos**. Este último control es importante: si la misma imagen cae a la vez en entrenamiento y en test, la métrica de test queda inflada (fuga de información).

```python
def build_index(class_root):
    rows, corrupt = [], []
    for class_dir in sorted(p for p in class_root.iterdir() if p.is_dir()):
        for f in sorted(class_dir.iterdir()):
            if f.suffix.lower() not in IMG_EXT:
                continue
            try:
                with Image.open(f) as im:
                    w, h = im.size
                    mode = im.mode
                    fmt = im.format
                    im.verify()
                if fmt not in ('JPEG', 'PNG', 'GIF', 'BMP'):      # formatos soportados por tf.io.decode_image
                    raise ValueError('formato no soportado por TensorFlow: %s' % fmt)
                md5 = hashlib.md5(f.read_bytes()).hexdigest()
                rows.append(dict(path=str(f), label=class_dir.name, width=w, height=h, mode=mode, md5=md5))
            except Exception as e:
                corrupt.append((str(f), repr(e)))
    return pd.DataFrame(rows), corrupt

df_all, corrupt = build_index(CLASS_ROOT)

print('Imagenes legibles :', len(df_all))
print('Ficheros corruptos:', len(corrupt))
print('Numero de clases  :', df_all['label'].nunique(), '->', sorted(df_all['label'].unique()))
print('Modos de color    :', df_all['mode'].value_counts().to_dict())

dup_mask = df_all.duplicated('md5', keep='first')
conflict = df_all.groupby('md5')['label'].nunique().gt(1).sum()
print('Duplicados exactos (se eliminan):', int(dup_mask.sum()))
print('Duplicados con etiquetas distintas:', int(conflict))
df = df_all[~dup_mask].reset_index(drop=True)
print('Imagenes tras eliminar duplicados:', len(df))
```

```python
if MODO_RAPIDO:
    df = df.groupby('label', group_keys=False).head(40).reset_index(drop=True)
    print('MODO_RAPIDO activo: se usan solo', len(df), 'imagenes')

CLASS_NAMES = sorted(df['label'].unique())
NUM_CLASSES = len(CLASS_NAMES)
df['label_id'] = df['label'].map({c: i for i, c in enumerate(CLASS_NAMES)}).astype('int32')

counts = df['label'].value_counts().reindex(CLASS_NAMES)
dist = pd.DataFrame({'imagenes': counts, 'porcentaje': (100 * counts / counts.sum()).round(1)})
display(dist)

imbalance = counts.max() / counts.min()
MAJORITY_BASELINE = counts.max() / counts.sum()
print('Ratio clase mayoritaria / minoritaria: %.2f' % imbalance)
print('Accuracy de un clasificador aleatorio uniforme : %.1f %%' % (100 / NUM_CLASSES))
print('Accuracy de predecir siempre la clase mayoritaria: %.1f %%' % (100 * MAJORITY_BASELINE))

fig, ax = plt.subplots(figsize=(7, 3.5))
sns.barplot(x=counts.index, y=counts.values, ax=ax, color='#4C72B0')
for i, v in enumerate(counts.values):
    ax.text(i, v, str(v), ha='center', va='bottom')
ax.set_title('Distribucion de clases'); ax.set_xlabel('clase'); ax.set_ylabel('numero de imagenes')
plt.show()
```

#### Análisis de resultados

## 1.4 Acondicionamiento y preprocesamiento

Decisiones adoptadas y motivo de cada una:

1. **Partición estratificada 70 % / 15 % / 15 %** (entrenamiento / validación / test). El dataset no trae una partición oficial. La estratificación conserva la proporción de clases en los tres subconjuntos. Los duplicados exactos se han eliminado antes de partir para evitar fugas.
   * *Entrenamiento*: ajuste de pesos.
   * *Validación*: monitorización, *early stopping* y **todas** las decisiones de diseño (arquitectura, hiperparámetros, regularización, elección de red preentrenada).
   * *Test*: se usa **una sola vez, al final**, para estimar el rendimiento de los modelos ya elegidos. Ninguna decisión se toma mirando test.
2. **Tamaño de entrada 224 x 224 x 3 para todos los modelos.** Es la resolución con la que se preentrenaron en ImageNet las arquitecturas de la estrategia 2. Usar la misma entrada en la estrategia 1 hace que la comparación entre estrategias no dependa de la resolución.
3. **Redimensionado conservando la relación de aspecto + recorte central.** El lado menor se escala a 224 px y se recorta el centro. Estirar la imagen directamente a un cuadrado deformaría la forma de los pétalos, que es precisamente una de las pistas discriminativas; el recorte central es razonable porque en este tipo de fotografía el motivo suele estar centrado.
4. **Normalización de intensidades dentro del modelo.** El *pipeline* entrega píxeles en [0, 255] y cada red incluye como primera capa su propia normalización: reescalado a [0, 1] en la CNN propia y el preprocesamiento específico de ImageNet en cada red preentrenada. Así todos los modelos consumen exactamente los mismos tensores y no es posible aplicar por error una normalización que no corresponde.
5. **Pipeline `tf.data` con caché en memoria.** Las imágenes se decodifican y redimensionan una sola vez y se guardan en RAM como `uint8`; después cada época solo baraja y agrupa en lotes. El *data augmentation* se aplica dentro del modelo (en GPU y solo en entrenamiento), de modo que la caché contiene siempre las imágenes originales.

```python
IMG_SIZE = 224
AUTOTUNE = tf.data.AUTOTUNE

train_df, tmp_df = train_test_split(df, test_size=0.30, stratify=df['label_id'], random_state=SEED)
val_df, test_df = train_test_split(tmp_df, test_size=0.50, stratify=tmp_df['label_id'], random_state=SEED)
train_df, val_df, test_df = [d.reset_index(drop=True) for d in (train_df, val_df, test_df)]

split_table = pd.DataFrame({
    'train': train_df['label'].value_counts(), 'val': val_df['label'].value_counts(),
    'test': test_df['label'].value_counts()}).reindex(CLASS_NAMES)
split_table.loc['TOTAL'] = split_table.sum()
display(split_table)

assert not set(train_df['md5']) & set(test_df['md5']) and not set(train_df['md5']) & set(val_df['md5'])
print('Sin solapamiento de imagenes entre particiones.')

# Incertidumbre de una accuracy medida en validacion (error estandar binomial, caso p = 0.8)
VAL_SE = float(np.sqrt(0.8 * 0.2 / len(val_df)))
print('Error estandar aproximado de la accuracy de validacion: +/- %.1f puntos' % (100 * VAL_SE))
```

```python
def load_image(path, label):
    img = tf.io.decode_image(tf.io.read_file(path), channels=3, expand_animations=False)
    img.set_shape([None, None, 3])
    hw = tf.cast(tf.shape(img)[:2], tf.float32)
    scale = IMG_SIZE / tf.reduce_min(hw)                       # el lado menor pasa a medir IMG_SIZE
    new_hw = tf.cast(tf.math.ceil(hw * scale), tf.int32)
    img = tf.image.resize(img, new_hw, method='bilinear', antialias=True)
    img = tf.image.resize_with_crop_or_pad(img, IMG_SIZE, IMG_SIZE)   # recorte central
    img = tf.cast(tf.clip_by_value(tf.round(img), 0.0, 255.0), tf.uint8)
    return img, label

def to_float(img, label):
    return tf.cast(img, tf.float32), label

def cached_base(frame):
    ds = tf.data.Dataset.from_tensor_slices((frame['path'].values, frame['label_id'].values))
    return ds.map(load_image, num_parallel_calls=AUTOTUNE).cache()

base_train, base_val, base_test = cached_base(train_df), cached_base(val_df), cached_base(test_df)

def batches(base, batch_size, training=False, n=None):
    ds = base
    if training:
        ds = ds.shuffle(n, seed=SEED, reshuffle_each_iteration=True)
    return ds.batch(batch_size).map(to_float, num_parallel_calls=AUTOTUNE).prefetch(AUTOTUNE)

def train_batches(batch_size):
    return batches(base_train, batch_size, training=True, n=len(train_df))

train_eval_ds = batches(base_train, 64)     # entrenamiento SIN barajar ni aumentar: para medir train en modo inferencia
val_ds = batches(base_val, 64)
test_ds = batches(base_test, 64)
y_test = test_df['label_id'].values

# Comprobacion del pipeline
xb, yb = next(iter(val_ds))
print('Lote:', xb.shape, xb.dtype, '| rango de pixeles: [%.0f, %.0f]' % (xb.numpy().min(), xb.numpy().max()))
fig, axes = plt.subplots(1, 8, figsize=(16, 2.4))
for i, ax in enumerate(axes):
    ax.imshow(xb[i].numpy().astype('uint8')); ax.set_title(CLASS_NAMES[int(yb[i])], fontsize=9); ax.axis('off')
plt.suptitle('Imagenes tal y como entran a los modelos (224x224, recorte central)', y=1.05)
plt.show()
```

# 2. Metodología experimental y utilidades comunes

Para que el proceso sea realmente experimental y no una sucesión de pruebas sueltas, se fijan **antes de entrenar nada** las reglas con las que se van a tomar las decisiones.

**Protocolo de cada experimento**

* Cada experimento cambia **un único factor** respecto a la configuración de referencia vigente (búsqueda progresiva o *greedy*). Así el efecto observado es atribuible a ese factor.
* Función de pérdida: entropía cruzada categórica (etiquetas enteras). Métrica: *accuracy*; es adecuada si las clases están razonablemente equilibradas (se comprobó en 1.3) y en test se complementa con *precision*, *recall* y F1 por clase.
* **Early stopping** sobre la pérdida de validación con restauración de los mejores pesos: cada modelo se evalúa en su mejor época y no en la última. El número de épocas es, por tanto, un máximo; la época en la que se detiene es un resultado más del experimento.
* Tras entrenar, se mide la *accuracy* de **entrenamiento en modo inferencia** (sin *dropout* ni *augmentation*) y la de validación con los pesos restaurados. La diferencia entre ambas (**gap**) es el indicador de sobreajuste. Se hace así porque la *accuracy* de entrenamiento que Keras muestra durante `fit` se calcula con la regularización activa y con pesos que cambian dentro de la época, y no es directamente comparable con la de validación.

**Regla de decisión (fijada de antemano)**

El conjunto de validación es pequeño, así que su *accuracy* tiene una incertidumbre de más de un punto (error estándar calculado arriba). No tiene sentido "elegir" por diferencias de décimas. Por ello:

* Un cambio se **adopta** si mejora la *accuracy* de validación en al menos **1 punto**.
* Si queda en **empate técnico** (diferencia menor de 1 punto), se adopta solo si reduce el *gap* de sobreajuste en al menos 2 puntos o si el modelo resultante es más simple (menos parámetros).
* En otro caso se **descarta** y se mantiene la referencia.

La comparación entre experimentos se hace con *accuracy* y no con la pérdida de validación porque esta última incluye el término de penalización L2 cuando se usa *weight regularization*, y dejaría de ser comparable entre configuraciones.

El conjunto de test no interviene en ninguna de estas decisiones.

```python
RESULTS, HISTORIES = {}, {}

EPOCHS_SCRATCH = 2 if MODO_RAPIDO else 40
PATIENCE_SCRATCH = 6
MARGIN, GAP_MARGIN = 1.0, 2.0          # puntos porcentuales de la regla de decision

def count_trainable(model):
    return int(sum(np.prod(w.shape) for w in model.trainable_weights))

def run_experiment(name, model, description, strategy, lr=1e-3, optimizer='adam', batch_size=32,
                   epochs=EPOCHS_SCRATCH, patience=PATIENCE_SCRATCH, reduce_lr=False):
    # Compila, entrena con early stopping, mide train/val en modo inferencia y registra el experimento.
    if optimizer == 'adam':
        opt = keras.optimizers.Adam(learning_rate=lr)
    elif optimizer == 'sgd':
        opt = keras.optimizers.SGD(learning_rate=lr, momentum=0.9, nesterov=True)
    else:
        raise ValueError(optimizer)
    model.compile(optimizer=opt, loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    callbacks = [keras.callbacks.EarlyStopping(monitor='val_loss', patience=patience, restore_best_weights=True)]
    if reduce_lr:
        callbacks.append(keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-6))

    print('=' * 100)
    print('EXPERIMENTO %s | %s' % (name, description))
    print('optimizador=%s  lr=%g  batch=%d  epocas_max=%d  params=%s (entrenables=%s)' % (
        optimizer, lr, batch_size, epochs, format(model.count_params(), ','), format(count_trainable(model), ',')))
    t0 = time.time()
    hist = model.fit(train_batches(batch_size), validation_data=val_ds, epochs=epochs,
                     callbacks=callbacks, verbose=2)
    secs = time.time() - t0

    h = {k: [float(v) for v in hist.history[k]] for k in ('loss', 'val_loss', 'accuracy', 'val_accuracy')}
    HISTORIES[name] = h
    _, tr_acc = model.evaluate(train_eval_ds, verbose=0)
    va_loss, va_acc = model.evaluate(val_ds, verbose=0)
    RESULTS[name] = dict(
        estrategia=strategy, descripcion=description, params=model.count_params(),
        params_entrenables=count_trainable(model), epocas=len(h['loss']),
        mejor_epoca=int(np.argmin(h['val_loss'])) + 1, seg_epoca=secs / len(h['loss']), tiempo_s=secs,
        train_acc=float(tr_acc), val_acc=float(va_acc), gap=float(tr_acc - va_acc), val_loss=float(va_loss))
    r = RESULTS[name]
    print('-> %s: train_acc=%.1f%%  val_acc=%.1f%%  gap=%.1f pts  mejor_epoca=%d/%d  (%.1f s/epoca)' % (
        name, 100 * r['train_acc'], 100 * r['val_acc'], 100 * r['gap'], r['mejor_epoca'], r['epocas'], r['seg_epoca']))
    plot_history(name)
    return model

def plot_history(names, title=None):
    # Curvas de perdida y accuracy. Si se pasan varias fases, se concatenan y se marca la frontera.
    names = [names] if isinstance(names, str) else list(names)
    h = {k: sum((HISTORIES[n][k] for n in names), []) for k in ('loss', 'val_loss', 'accuracy', 'val_accuracy')}
    ep = np.arange(1, len(h['loss']) + 1)
    fig, ax = plt.subplots(1, 2, figsize=(12, 3.6))
    ax[0].plot(ep, h['loss'], label='entrenamiento'); ax[0].plot(ep, h['val_loss'], label='validacion')
    ax[0].set_title('Perdida'); ax[0].set_xlabel('epoca'); ax[0].legend()
    ax[1].plot(ep, h['accuracy'], label='entrenamiento'); ax[1].plot(ep, h['val_accuracy'], label='validacion')
    ax[1].set_title('Accuracy'); ax[1].set_xlabel('epoca'); ax[1].legend()
    if len(names) == 1:
        best = int(np.argmin(h['val_loss'])) + 1
        for a in ax:
            a.axvline(best, color='gray', ls='--', lw=1)
    else:
        b = 0
        for n in names[:-1]:
            b += len(HISTORIES[n]['loss'])
            for a in ax:
                a.axvline(b + 0.5, color='k', ls=':', lw=1)
    fig.suptitle(title or ' + '.join(names), y=1.03)
    plt.tight_layout(); plt.show()

def summary(names):
    cols = ['descripcion', 'params', 'epocas', 'mejor_epoca', 'seg_epoca', 'train_acc', 'val_acc', 'gap']
    t = pd.DataFrame({n: RESULTS[n] for n in names}).T[cols].copy()
    for c in ('train_acc', 'val_acc', 'gap'):
        t[c] = (100 * t[c].astype(float)).round(1)
    t['seg_epoca'] = t['seg_epoca'].astype(float).round(1)
    return t

def compare(names, title):
    # Tabla resumen y curvas de validacion superpuestas de varios experimentos.
    with pd.option_context('display.max_colwidth', 60, 'display.width', 200):
        display(summary(names))
    fig, ax = plt.subplots(1, 2, figsize=(12, 3.6))
    for n in names:
        ep = np.arange(1, len(HISTORIES[n]['val_loss']) + 1)
        ax[0].plot(ep, HISTORIES[n]['val_loss'], label=n)
        ax[1].plot(ep, HISTORIES[n]['val_accuracy'], label=n)
    ax[0].set_title('Perdida de validacion'); ax[1].set_title('Accuracy de validacion')
    for a in ax:
        a.set_xlabel('epoca'); a.legend(fontsize=8)
    fig.suptitle(title, y=1.03)
    plt.tight_layout(); plt.show()

def decide(candidates, reference, prefer_on_tie=False):
    # Aplica la regla de decision fijada de antemano. Devuelve el nombre del experimento ganador.
    candidates = [candidates] if isinstance(candidates, str) else list(candidates)
    best = max(candidates, key=lambda n: RESULTS[n]['val_acc'])
    c, r = RESULTS[best], RESULTS[reference]
    d_acc, d_gap = 100 * (c['val_acc'] - r['val_acc']), 100 * (c['gap'] - r['gap'])
    if d_acc >= MARGIN:
        winner, why = best, 'mejora clara en validacion'
    elif d_acc > -MARGIN and d_gap <= -GAP_MARGIN:
        winner, why = best, 'empate tecnico en validacion, pero reduce el sobreajuste'
    elif d_acc > -MARGIN and prefer_on_tie:
        winner, why = best, 'empate tecnico en validacion y modelo preferible por simplicidad'
    else:
        winner, why = reference, 'el mejor candidato no supera la regla de decision'
    print('DECISION | referencia=%s (val %.1f%%, gap %.1f) | mejor candidato=%s (val %.1f%%, gap %.1f)' % (
        reference, 100 * r['val_acc'], 100 * r['gap'], best, 100 * c['val_acc'], 100 * c['gap']))
    print('         | dif. val_acc = %+.1f pts, dif. gap = %+.1f pts -> SE ADOPTA: %s (%s)' % (d_acc, d_gap, winner, why))
    return winner
```

# 3. Estrategia 1: CNN entrenada desde cero (*from scratch*)

## 3.1 Constructor parametrizable de la arquitectura

Todas las variantes de la CNN se generan con una única función. Cada decisión de diseño es un parámetro, lo que garantiza que dos experimentos solo difieren en el factor que se está estudiando.

Estructura general: bloques `[Conv2D -> (BatchNorm) -> activación] x n -> MaxPooling2D`, con un número de filtros creciente (al reducirse la resolución espacial se aumenta el número de canales para que la red pueda representar patrones cada vez más abstractos), seguidos de un clasificador (`Flatten` o `GlobalAveragePooling2D`, una capa densa y la salida `softmax` con una neurona por clase).

El *data augmentation* se define como un bloque de capas de Keras que solo actúa en entrenamiento. Las transformaciones elegidas son las que conservan la etiqueta en este problema: volteo horizontal, pequeñas rotaciones, zoom, traslación y cambio de contraste. No se usa volteo vertical ni rotaciones grandes porque las flores se fotografían casi siempre con una orientación natural, y generar imágenes que no aparecerán en validación ni en test no ayuda.

```python
def build_augmenter():
    return keras.Sequential([
        layers.RandomFlip('horizontal'),
        layers.RandomRotation(0.08),
        layers.RandomZoom(0.15),
        layers.RandomTranslation(0.10, 0.10),
        layers.RandomContrast(0.15),
    ], name='data_augmentation')

def build_cnn(filters=(32, 64, 128), kernel_size=3, convs_per_block=1, activation='relu',
              batch_norm=False, head='flatten', dense_units=128, dropout=0.0, l2=0.0, augment=False):
    keras.utils.set_random_seed(SEED)
    reg = regularizers.l2(l2) if l2 > 0 else None
    inputs = keras.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
    x = inputs
    if augment:
        x = build_augmenter()(x)
    x = layers.Rescaling(1.0 / 255)(x)                           # normalizacion a [0, 1]
    for f in filters:
        for _ in range(convs_per_block):
            x = layers.Conv2D(f, kernel_size, padding='same', use_bias=not batch_norm, kernel_regularizer=reg)(x)
            if batch_norm:
                x = layers.BatchNormalization()(x)
            x = layers.Activation(activation)(x)
        x = layers.MaxPooling2D(2)(x)
    x = layers.Flatten()(x) if head == 'flatten' else layers.GlobalAveragePooling2D()(x)
    if dropout > 0:
        x = layers.Dropout(dropout)(x)
    x = layers.Dense(dense_units, activation=activation, kernel_regularizer=reg)(x)
    if dropout > 0:
        x = layers.Dropout(dropout)(x)
    outputs = layers.Dense(NUM_CLASSES, activation='softmax')(x)
    return keras.Model(inputs, outputs)

# Configuracion vigente (se ira actualizando con las decisiones) y registro de cambios de cada experimento
MODEL_CFG = dict(filters=(32, 64, 128), kernel_size=3, convs_per_block=1, activation='relu',
                 batch_norm=False, head='flatten', dense_units=128, dropout=0.0, l2=0.0, augment=False)
TRAIN_CFG = dict(lr=1e-3, optimizer='adam', batch_size=32)
EXP_CHANGES = {}

def scratch_experiment(name, description, model_changes=None, train_changes=None, **fit_kwargs):
    # Entrena la configuracion vigente modificando unicamente los factores indicados.
    model_changes, train_changes = model_changes or {}, train_changes or {}
    EXP_CHANGES[name] = (model_changes, train_changes)
    model = build_cnn(**{**MODEL_CFG, **model_changes})
    run_experiment(name, model, description, 'scratch', **{**TRAIN_CFG, **train_changes}, **fit_kwargs)
    del model; gc.collect()

def adopt(winner):
    # Incorpora a la configuracion vigente los cambios del experimento ganador.
    mc, tc = EXP_CHANGES[winner]
    MODEL_CFG.update(mc); TRAIN_CFG.update(tc)
    print('Configuracion vigente -> modelo:', MODEL_CFG)
    print('                      -> entrenamiento:', TRAIN_CFG)
    return winner
```

## 3.2 Experimento S0: modelo base sin regularización

**Objetivo.** Establecer un punto de partida y diagnosticar el problema. Se parte de una CNN sencilla: 3 bloques convolucionales (32, 64 y 128 filtros de 3x3, ReLU, *max pooling* 2x2), `Flatten`, una capa densa de 128 neuronas y la salida. Optimizador Adam con su tasa de aprendizaje por defecto (0,001) y lotes de 32. Sin ninguna regularización.

**Hipótesis.** Tras tres reducciones a la mitad, el mapa de características mide 28x28x128, de modo que la capa `Flatten -> Dense` concentra más de 12 millones de pesos para unas 3.000 imágenes de entrenamiento. Se espera que el modelo memorice el conjunto de entrenamiento: *accuracy* de entrenamiento cercana al 100 %, pérdida de validación que alcanza un mínimo en pocas épocas y después sube (sobreajuste). Aun así debe superar con claridad las referencias triviales.

```python
build_cnn(**MODEL_CFG).summary()
scratch_experiment('S0_base', 'CNN base: 3 bloques, Flatten, sin regularizacion')
REF = 'S0_base'
print('Referencias triviales: azar %.1f%% | clase mayoritaria %.1f%%' % (100 / NUM_CLASSES, 100 * MAJORITY_BASELINE))
```

#### Análisis de resultados


* **Se confirma el sobreajuste.** La pérdida de validación alcanza su mínimo en la **época 2** (0,942) y a partir de ahí sube sin parar, hasta 2,121 en la época 8, mientras que la *accuracy* de entrenamiento durante `fit` pasa del 44 % al **98,5 %**. Es el patrón típico de memorización: el modelo, con 12,9 millones de parámetros (el 99 % en la capa `Flatten -> Dense`), aprende de memoria las 3.017 imágenes de entrenamiento.
* **El *early stopping* oculta parte del problema.** Al restaurar los pesos de la época 2, el *gap* medido en modo inferencia es de 10,5 puntos (73,7 % frente a 63,2 %). Sin esa restauración, en la época 8 la diferencia superaría los 35 puntos. El modelo apenas puede entrenar dos épocas antes de empezar a empeorar.
* **Aun así, aprende.** Con un 63,2 % en validación supera ampliamente a las referencias triviales (24,4 %), así que extrae patrones útiles.
* **Diagnóstico y siguientes pasos.** El problema principal es la **sobreparametrización de la cabeza densa** y una **tasa de aprendizaje que hace que el modelo memorice en muy pocas épocas**. Eso motiva el orden de los experimentos: primero la tasa de aprendizaje (S1), luego la arquitectura (S3 y S4, que atacan el exceso de parámetros) y después la regularización (S6 a S9).

## 3.3 Experimento S1: optimizador y tasa de aprendizaje

**Motivación.** La tasa de aprendizaje es el hiperparámetro que más condiciona el entrenamiento, y conviene fijarla antes de tocar la arquitectura para no atribuir a la arquitectura un problema de optimización. Se compara la referencia (Adam, 0,001) con una tasa diez veces mayor, otra diez veces menor, y SGD con momento de Nesterov (0,9) a 0,01.

**Hipótesis.** Con 0,01, Adam debería ser inestable o quedarse estancado cerca del azar; con 0,0001 debería aprender de forma más lenta y suave, quizá sin llegar a converger en las épocas disponibles; SGD con momento suele converger más despacio que Adam pero a veces generaliza algo mejor.

```python
scratch_experiment('S1_adam_lr1e-2', 'Adam, lr = 0.01', train_changes=dict(lr=1e-2))
scratch_experiment('S1_adam_lr1e-4', 'Adam, lr = 0.0001', train_changes=dict(lr=1e-4))
scratch_experiment('S1_sgd_lr1e-2', 'SGD + momento Nesterov, lr = 0.01', train_changes=dict(optimizer='sgd', lr=1e-2))
cands = ['S1_adam_lr1e-2', 'S1_adam_lr1e-4', 'S1_sgd_lr1e-2']
compare([REF] + cands, 'S1: optimizador y tasa de aprendizaje')
REF = adopt(decide(cands, REF))
```

#### Análisis de resultados


* **Adam con 0,01: colapso.** Tras la primera época (pérdida de 3,42) la pérdida se queda clavada en 1,60, que coincide con la entropía de la distribución de clases del dataset. Es decir, la red ha dejado de usar la imagen y devuelve siempre las proporciones de cada clase: predice siempre *dandelion*, y por eso su *accuracy* es exactamente el 24,4 % de la clase mayoritaria. Lo más probable es que el primer paso, demasiado grande, haya llevado las neuronas ReLU a una zona en la que dejan de activarse y ya no se recuperan. Se confirma la hipótesis de inestabilidad.
* **Adam con 0,0001: mejor validación.** Aprende de forma más gradual: alcanza su mejor pérdida de validación en la época 8 (0,878, frente a 0,942 de la referencia) y logra un **66,5 %** en validación (+3,2 puntos). A cambio, el *gap* sube a 21,1 puntos, porque al entrenar más épocas antes de detenerse, el modelo también memoriza más.
* **SGD con momento a 0,01.** Converge de forma más irregular: la pérdida de validación salta de una época a otra (0,94 en la época 4 y 1,19 en la 5) y a partir de ahí crece hasta 1,87. La validación se queda en un 62,6 %, sin mejorar a Adam. No se observa la mejor generalización que a veces se atribuye a SGD.
* **Decisión: Adam con 0,0001** (+3,2 puntos, por encima del margen de 1 punto). El aumento del *gap* indica que el sobreajuste sigue siendo el problema principal; se aborda en los experimentos de arquitectura y regularización.

## 3.4 Experimento S2: tamaño del lote

**Motivación.** El tamaño del lote determina el ruido del gradiente y el número de actualizaciones por época. Lotes pequeños dan gradientes más ruidosos (cierto efecto regularizador) y más actualizaciones, a costa de épocas más lentas; lotes grandes dan gradientes más estables y épocas más rápidas, pero menos actualizaciones. Se compara la referencia (32) con 16 y 64, manteniendo la tasa de aprendizaje.

**Hipótesis.** En un dataset pequeño no se esperan diferencias grandes; si las hay, deberían favorecer a los lotes pequeños o medianos.

```python
scratch_experiment('S2_batch16', 'Lote de 16', train_changes=dict(batch_size=16))
scratch_experiment('S2_batch64', 'Lote de 64', train_changes=dict(batch_size=64))
cands = ['S2_batch16', 'S2_batch64']
compare([REF] + cands, 'S2: tamano del lote')
REF = adopt(decide(cands, REF))
```

#### Análisis de resultados


* **Ningún tamaño mejora al de 32.** Con lotes de 16 la validación baja a 64,5 % (-2,0 puntos) y con lotes de 64, a 64,1 % (-2,4 puntos). Las diferencias son similares a la incertidumbre de la validación (±1,6 puntos), así que lo más que puede afirmarse es que el lote no es un factor determinante en este problema.
* **El efecto sobre el sobreajuste sí sigue la teoría.** El lote de 16 tiene el menor *gap* (12,2 puntos) y el de 64 el mayor (17,2). Los gradientes más ruidosos de los lotes pequeños actúan como una regularización suave, pero aquí no se traduce en mejor validación.
* **Coste.** En CPU las diferencias de tiempo son pequeñas (de 46 a 53 s por época). El lote de 16 es algo más lento por el mayor número de actualizaciones.
* **Decisión:** se mantiene el **lote de 32**, de acuerdo con la regla de decisión.

## 3.5 Experimento S3: profundidad de la red (número de capas convolucionales y de filtros)

**Motivación.** Con 3 bloques, cada neurona de la última capa convolucional solo "ve" una región pequeña de una imagen de 224x224 (campo receptivo reducido) y el mapa final sigue siendo muy grande. Añadir bloques aumenta el campo receptivo y el nivel de abstracción, y además reduce el tamaño del mapa que llega al clasificador. Se prueban 4 bloques (32-64-128-256) y 5 bloques (32-64-128-256-512).

**Hipótesis.** Más profundidad debería mejorar la validación: las características son más ricas y, paradójicamente, el modelo tiene menos parámetros en total porque la capa densa tras `Flatten` se reduce mucho (de 28x28x128 a 7x7x512 entradas).

```python
scratch_experiment('S3_4bloques', '4 bloques conv (32-64-128-256)', model_changes=dict(filters=(32, 64, 128, 256)))
scratch_experiment('S3_5bloques', '5 bloques conv (32-64-128-256-512)', model_changes=dict(filters=(32, 64, 128, 256, 512)))
cands = ['S3_4bloques', 'S3_5bloques']
compare([REF] + cands, 'S3: profundidad')
REF = adopt(decide(cands, REF))
```

#### Análisis de resultados


* **Más profundidad con menos parámetros.** Cada bloque adicional reduce a la mitad el mapa que llega a `Flatten` y, con él, el tamaño de la capa densa. El modelo pasa de 12,9 millones de parámetros (3 bloques) a 6,8 millones (4 bloques) y a 4,8 millones (5 bloques).
* **4 bloques: -2,5 puntos.** Llega a 64,0 % frente al 66,5 % de la referencia. Con 14x14x256 entradas a la capa densa, la cabeza sigue siendo enorme y el modelo continúa sobreajustando (*gap* de 14,1).
* **5 bloques: +2,3 puntos.** Llega al **68,8 %** y reduce el *gap* de 21,1 a 12,1 puntos. El campo receptivo de la última capa ya cubre buena parte de la flor y el modelo combina patrones de más alto nivel. Es la primera mejora que llega a la vez por las dos vías: más validación y menos sobreajuste.
* **Decisión: 5 bloques (32-64-128-256-512).** Se confirma la hipótesis, aunque solo con 5 bloques, no con 4. Aun así, de los 4,8 millones de parámetros, 3,2 siguen en la capa densa tras `Flatten`, lo que motiva el siguiente experimento.

## 3.6 Experimento S4: capas finales, `Flatten` frente a `GlobalAveragePooling2D`

**Motivación.** `Flatten` conserva la posición espacial de cada activación y obliga a la capa densa a aprender un peso por posición y canal; es la parte del modelo con más parámetros y la principal sospechosa del sobreajuste. `GlobalAveragePooling2D` (GAP) promedia cada canal y entrega un vector de tantos valores como filtros tenga el último bloque: reduce drásticamente los parámetros y hace al modelo más invariante a la posición de la flor.

**Hipótesis.** GAP debería reducir el *gap* de sobreajuste con una *accuracy* de validación igual o mejor. Como también produce un modelo mucho más pequeño, en caso de empate técnico se prefiere GAP.

```python
scratch_experiment('S4_gap', 'Clasificador con GlobalAveragePooling en lugar de Flatten', model_changes=dict(head='gap'))
compare([REF, 'S4_gap'], 'S4: Flatten frente a GAP')
REF = adopt(decide('S4_gap', REF, prefer_on_tie=True))
```

#### Análisis de resultados


* **GAP mejora en todo.** Sustituir `Flatten` por `GlobalAveragePooling2D` reduce los parámetros de 4,78 a **1,63 millones** (-66 %), sube la validación a **72,5 %** (+3,7 puntos) y baja el *gap* de 12,1 a **4,8 puntos**. Se confirma la hipótesis: la capa densa conectada a `Flatten` era la principal fuente de sobreajuste, y promediar cada canal hace al modelo menos dependiente de dónde aparece la flor.
* **Nueva limitación: el número de épocas.** El modelo alcanza su mejor época en la **34 de 40**. Al tener menos capacidad de memorizar, aprende más despacio y sigue mejorando casi hasta el límite de épocas. A partir de aquí, el máximo de 40 épocas empieza a condicionar las comparaciones; se tiene en cuenta al interpretar los experimentos siguientes.
* **Decisión: cabeza GAP**, por mejora clara en validación.

## 3.7 Experimento S5: tamaño del *kernel*, convoluciones por bloque y función de activación

Tres decisiones de diseño de las capas convolucionales, evaluadas cada una por separado frente a la referencia vigente:

* **Kernel 5x5 en lugar de 3x3.** Mayor campo receptivo por capa, pero casi el triple de parámetros por filtro (25 frente a 9 pesos por canal de entrada).
* **Dos convoluciones 3x3 por bloque (estilo VGG).** Dos capas 3x3 apiladas cubren el mismo campo receptivo que una 5x5 con menos parámetros y una no linealidad adicional. Hipótesis: es la forma más eficiente de ganar capacidad.
* **Activación ELU en lugar de ReLU.** ELU no anula las activaciones negativas, lo que evita neuronas "muertas" y centra las activaciones. Hipótesis: en una red poco profunda la diferencia con ReLU será pequeña.

```python
scratch_experiment('S5_kernel5', 'Kernels de 5x5', model_changes=dict(kernel_size=5))
scratch_experiment('S5_2conv', 'Dos convoluciones 3x3 por bloque', model_changes=dict(convs_per_block=2))
scratch_experiment('S5_elu', 'Activacion ELU', model_changes=dict(activation='elu'))
cands = ['S5_kernel5', 'S5_2conv', 'S5_elu']
compare([REF] + cands, 'S5: kernel, convoluciones por bloque y activacion')
REF = adopt(decide(cands, REF))
```

#### Análisis de resultados


Las tres variantes mejoran a la referencia (72,5 %), lo que indica que el modelo con GAP aún tenía margen para ganar capacidad:

* **Kernel 5x5: 79,9 % (+7,4 puntos).** Es la mejora individual más grande de toda la estrategia 1. Un campo receptivo mayor en cada capa ayuda a captar la estructura de los pétalos alrededor del centro de la flor. A cambio, los parámetros suben a 4,42 millones, el tiempo por época se duplica (de 59 a 122 s) y el *gap* aumenta a 11,7 puntos. Su mejor época es la 39 de 40: sigue al límite del presupuesto de épocas.
* **Dos convoluciones 3x3 por bloque: 77,4 % (+4,9 puntos)** con un *gap* mucho menor (5,3 puntos), pero es la variante más cara (179 s por época). La hipótesis de que sería la forma más eficiente de ganar capacidad no se confirma aquí: con el mismo número de épocas, el *kernel* 5x5 obtiene más validación.
* **ELU: 74,5 % (+2,0 puntos)** con el *gap* más bajo (3,6 puntos) y sin coste adicional. Es una mejora pequeña, como se preveía para una red poco profunda.
* **Decisión: kernel 5x5**, por ser el de mayor *accuracy* de validación. Hay que matizar que la variante de dos convoluciones generaliza mejor (*gap* de 5,3 frente a 11,7) y sería una alternativa razonable si se priorizara reducir el sobreajuste; la regla de decisión, fijada de antemano, prioriza la validación.

## 3.8 Experimentos de regularización

Fijada la arquitectura, se abordan de forma explícita los problemas detectados, con las cuatro técnicas estudiadas en la asignatura. Se evalúan **de forma acumulativa**: cada técnica se prueba sobre la configuración vigente y solo se incorpora si supera la regla de decisión.

### S6: *Batch normalization*

**Problema que aborda.** No es un regularizador en sentido estricto: su función principal es estabilizar y acelerar la optimización normalizando las activaciones de cada capa con las estadísticas del lote, lo que permite un entrenamiento menos sensible a la inicialización y a la tasa de aprendizaje. Como efecto secundario introduce ruido (las estadísticas cambian de un lote a otro) con un leve efecto regularizador. Se coloca entre la convolución y la activación, y la convolución se define sin sesgo porque BN ya incluye un desplazamiento aprendible.

**Hipótesis.** Convergencia más rápida (mejor época más temprana) y, probablemente, mejor validación. Es habitual que la curva de validación sea más ruidosa en las primeras épocas, mientras las medias móviles de BN aún no se han estabilizado.

```python
scratch_experiment('S6_batchnorm', 'Batch normalization tras cada convolucion', model_changes=dict(batch_norm=True))
compare([REF, 'S6_batchnorm'], 'S6: batch normalization')
REF = adopt(decide('S6_batchnorm', REF))
```

#### Análisis de resultados


* **Acelera el ajuste a entrenamiento.** En la primera época, la *accuracy* de entrenamiento es del 54,9 % frente al 35,2 % sin BN. Se confirma el efecto principal esperado: la optimización es más rápida.
* **Validación muy inestable.** En las épocas 1 y 2 la validación es del 24,4 % y 24,7 %: el modelo predice casi siempre una sola clase. En las épocas siguientes salta de forma irregular (por ejemplo, de 65,7 % a 74,8 % y de nuevo a 73,4 %). Es el comportamiento anticipado: en validación BN usa medias móviles que aún no reflejan bien las estadísticas reales de las activaciones. Esta inestabilidad dispara el *early stopping*, que detiene el entrenamiento en la **época 17** (mejor época, la 11), cuando la *accuracy* de entrenamiento todavía subía.
* **Resultado:** 75,6 % en validación (-4,3 puntos) con un *gap* de 7,1 (-4,6 puntos). Además, es la variante más lenta (184 s por época).
* **Decisión: se descarta BN** según la regla. Hay que interpretarlo con cautela: con una paciencia mayor, o con BN combinada con una reducción de la tasa de aprendizaje, el resultado podría ser distinto. Lo que sí muestra el experimento es que, con este protocolo, BN no compensa.

### S7: *Dropout*

**Problema que aborda.** El sobreajuste por co-adaptación de neuronas: al desactivar aleatoriamente una fracción de las activaciones en cada paso, la red no puede depender de combinaciones concretas de neuronas y aprende representaciones redundantes; equivale de forma aproximada a promediar muchas subredes. Se aplica en el clasificador (antes y después de la capa densa), que es donde se concentra la capacidad de memorización. Se prueban tasas de 0,3 y 0,5.

**Hipótesis.** Reducción del *gap*; la *accuracy* de entrenamiento durante `fit` bajará (se mide con *dropout* activo) y la convergencia será algo más lenta. Una tasa demasiado alta puede provocar subajuste.

```python
scratch_experiment('S7_dropout0.3', 'Dropout 0.3 en el clasificador', model_changes=dict(dropout=0.3))
scratch_experiment('S7_dropout0.5', 'Dropout 0.5 en el clasificador', model_changes=dict(dropout=0.5))
cands = ['S7_dropout0.3', 'S7_dropout0.5']
compare([REF] + cands, 'S7: dropout')
REF = adopt(decide(cands, REF))
```

#### Análisis de resultados


* **Menos sobreajuste, menos validación.** El *dropout* reduce el *gap* de 11,7 a 6,6 puntos (tasa 0,3) y a 7,2 (tasa 0,5), pero la validación baja a 76,7 % y 77,1 %. Ambas variantes se detienen por *early stopping* antes del límite (épocas 33 y 34), así que no es un problema de épocas: el entrenamiento ha terminado.
* **Por qué el efecto es limitado.** Con la cabeza GAP, la parte densa del modelo es pequeña (unos 66.000 pesos). El *dropout* regulariza un componente que ya no es la principal fuente de sobreajuste, que ahora está en los filtros convolucionales.
* **Matiz sobre la referencia.** El 79,9 % de la referencia corresponde a una época especialmente buena: en las épocas 35 a 40 su validación osciló entre 76,0 % y 79,9 %. Comparadas con ese rango, las dos tasas de *dropout* están en el mismo nivel.
* **Decisión: no se adopta el dropout** (-2,8 puntos en el mejor caso, más allá del margen de 1 punto).

### S8: *Weight regularization* (L2)

**Problema que aborda.** Pesos de gran magnitud, asociados a funciones muy sensibles a pequeños cambios de la entrada. La penalización L2 añade a la pérdida un término proporcional a la suma de los cuadrados de los pesos, empujándolos hacia valores pequeños y soluciones más suaves. Se aplica a los *kernels* de las capas convolucionales y de la capa densa, con coeficientes 0,0001 y 0,001.

**Nota metodológica.** La pérdida que muestra Keras incluye el término de penalización, así que las curvas de pérdida de estos experimentos parten de valores más altos y no son comparables en valor absoluto con las de la referencia; la comparación se hace con la *accuracy*.

**Hipótesis.** Efecto moderado. Si el modelo ya usa BN, el efecto de L2 es menos directo, porque BN hace a la red invariante a la escala de los pesos de la convolución previa; un coeficiente alto puede causar subajuste.

```python
scratch_experiment('S8_l2_1e-4', 'Regularizacion L2 = 0.0001', model_changes=dict(l2=1e-4))
scratch_experiment('S8_l2_1e-3', 'Regularizacion L2 = 0.001', model_changes=dict(l2=1e-3))
cands = ['S8_l2_1e-4', 'S8_l2_1e-3']
compare([REF] + cands, 'S8: weight regularization L2')
REF = adopt(decide(cands, REF))
```

#### Análisis de resultados


* **L2 = 0,0001:** 77,6 % de validación (-2,3 puntos) y *gap* de 6,8 (-4,9). Se detiene en la época 35, así que ha convergido. Reduce el sobreajuste sin llegar a compensar en validación.
* **L2 = 0,001:** 74,0 % (-5,9 puntos) y *gap* de 4,8. La *accuracy* de entrenamiento cae a 78,9 % (frente al 91,6 % de la referencia): el modelo tiene **subajuste**, porque la penalización es tan fuerte que le impide ajustar bien incluso los datos de entrenamiento. Además, su mejor época es la 34 de 40, de modo que también aprende más despacio.
* **Decisión: no se adopta L2.** El coeficiente pequeño se queda cerca de la referencia (dentro del rango de oscilación descrito en S7) y el grande perjudica claramente. Esto encaja con la hipótesis: el efecto de L2 es moderado y, con un coeficiente alto, provoca subajuste.

### S9: *Data augmentation*

**Problema que aborda.** La causa raíz del sobreajuste: la escasez de datos. En lugar de restringir el modelo, se amplía artificialmente la variabilidad del conjunto de entrenamiento con transformaciones que conservan la etiqueta, de modo que la red no ve dos veces exactamente la misma imagen y se le fuerza a aprender invarianzas (a la posición, la escala, la orientación y el contraste). Primero se visualiza el efecto de las transformaciones definidas en 3.1.

**Hipótesis.** Debería ser la técnica con mayor efecto sobre el *gap*. El entrenamiento será más lento en converger (la tarea de entrenamiento se vuelve más difícil), y es posible que la *accuracy* de entrenamiento durante `fit` quede por debajo de la de validación, porque esta última se calcula sobre imágenes sin transformar.

```python
aug = build_augmenter()
sample = tf.cast(next(iter(base_train.batch(1)))[0], tf.float32)
fig, axes = plt.subplots(1, 8, figsize=(16, 2.4))
axes[0].imshow(sample[0].numpy().astype('uint8')); axes[0].set_title('original'); axes[0].axis('off')
for ax in axes[1:]:
    out = aug(sample, training=True)[0].numpy()
    ax.imshow(np.clip(out, 0, 255).astype('uint8')); ax.axis('off')
plt.suptitle('Una imagen de entrenamiento y siete versiones aumentadas', y=1.05); plt.show()

scratch_experiment('S9_augment', 'Data augmentation (flip, rotacion, zoom, traslacion, contraste)',
                   model_changes=dict(augment=True))
compare([REF, 'S9_augment'], 'S9: data augmentation')
REF = adopt(decide('S9_augment', REF))
```

#### Análisis de resultados


* **Reduce el sobreajuste:** el *gap* baja de 11,7 a 7,2 puntos.
* **Aprende más despacio.** En la época 3 la validación es del 53,5 % frente al 60,9 % sin aumentar, y la mejor época es la **37 de 40**: el modelo seguía mejorando al llegar al límite. También se observa lo anticipado en la hipótesis: durante `fit` la *accuracy* de entrenamiento (81,5 % en la última época, con imágenes transformadas) queda cerca de la de validación.
* **Resultado:** 76,8 % en validación (-3,1 puntos). **No se adopta** según la regla.
* **Comparación con las otras técnicas.** La hipótesis de que sería la técnica con mayor efecto sobre el *gap* **no se confirma**: las cuatro técnicas reducen el *gap* entre 4,5 y 6,9 puntos (BN: -4,6; *dropout*: -4,5 a -5,1; L2: -4,9 a -6,9; *augmentation*: -4,5) y todas pierden entre 2,3 y 5,9 puntos de validación.

**Conclusión sobre la regularización en la estrategia 1.** Todas las técnicas hacen lo que se espera de ellas, que es reducir la distancia entre entrenamiento y validación, pero ninguna mejora la validación en estas condiciones. Hay dos razones. Primero, la referencia ya generaliza razonablemente gracias a decisiones de arquitectura (GAP y cinco bloques), que se han revelado como la forma más eficaz de controlar el sobreajuste. Segundo, el presupuesto fijo de 40 épocas favorece a los modelos que aprenden rápido, y las técnicas de regularización, por diseño, ralentizan el aprendizaje. Es especialmente claro en el *augmentation* (mejor época 37 de 40). Una comparación más concluyente exigiría entrenar cada variante hasta su convergencia; se recoge como limitación en las conclusiones.

## 3.9 Resumen de la experimentación y modelo final de la estrategia 1

La tabla recoge todos los experimentos de la estrategia 1 en orden. A continuación se entrena el **modelo final** con la configuración resultante de las decisiones anteriores y un presupuesto de entrenamiento mayor: más épocas, más paciencia en el *early stopping* y **reducción de la tasa de aprendizaje cuando la pérdida de validación se estanca** (`ReduceLROnPlateau`, factor 0,5). La idea es usar pasos grandes al principio para avanzar rápido y pasos pequeños al final para afinar; con una tasa fija, las últimas épocas suelen oscilar alrededor del mínimo.

```python
scratch_names = [n for n in RESULTS if RESULTS[n]['estrategia'] == 'scratch']
with pd.option_context('display.max_colwidth', 60, 'display.width', 200):
    display(summary(scratch_names))

fig, ax = plt.subplots(figsize=(11, 4))
t = summary(scratch_names)
x = np.arange(len(t))
ax.bar(x - 0.2, t['train_acc'].astype(float), 0.4, label='train (modo inferencia)')
ax.bar(x + 0.2, t['val_acc'].astype(float), 0.4, label='validacion')
ax.axhline(100 * MAJORITY_BASELINE, color='k', ls='--', lw=1, label='clase mayoritaria')
ax.set_xticks(x); ax.set_xticklabels(t.index, rotation=60, ha='right', fontsize=8)
ax.set_ylabel('accuracy (%)'); ax.set_title('Estrategia 1: evolucion de los experimentos'); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()

print('Configuracion final -> modelo:', MODEL_CFG)
print('                    -> entrenamiento:', TRAIN_CFG)
```

# 4. Estrategia 2: redes preentrenadas en ImageNet

## 4.1 Planteamiento

Se reutilizan redes cuyos pesos se han aprendido sobre ImageNet (1,28 millones de imágenes, 1.000 clases). Las primeras capas de una CNN aprenden detectores genéricos (bordes, texturas, colores) y las últimas, patrones cada vez más específicos de la tarea original. La estrategia tiene dos fases:

1. **Transfer learning (extracción de características).** Se elimina el clasificador original (`include_top=False`), se **congela toda la base convolucional** y se entrena únicamente una cabeza nueva: `GlobalAveragePooling2D -> Dropout -> Dense(softmax)`. Solo se ajustan unos pocos miles de pesos, por lo que el riesgo de sobreajuste es bajo y se puede usar una tasa de aprendizaje normal (0,001).
2. **Fine-tuning.** Con la cabeza ya entrenada, se **descongela el último bloque convolucional** y se continúa el entrenamiento con una **tasa de aprendizaje muy baja (0,00001)**, para adaptar las características de más alto nivel al nuevo dominio sin destruir lo aprendido. El orden es importante: si se descongelara la base con una cabeza recién inicializada, los gradientes grandes y aleatorios de esa cabeza degradarían los pesos preentrenados.

**Arquitecturas comparadas** (tres familias de diseño distintas):

| Arquitectura | Idea principal | Preprocesamiento de entrada |
|---|---|---|
| **VGG16** | Apilamiento secuencial de convoluciones 3x3; sin *batch normalization* | RGB a BGR y resta de la media de ImageNet por canal |
| **ResNet50** | Conexiones residuales, que permiten entrenar redes mucho más profundas | Igual que VGG16 |
| **MobileNetV2** | Convoluciones separables en profundidad y bloques residuales invertidos; muy ligera | Reescalado a [-1, 1] |

Cada red exige el mismo preprocesamiento con el que fue entrenada; se implementa como primera capa del modelo.

**Tratamiento de *batch normalization* durante el *fine-tuning*.** En ResNet50 y MobileNetV2, la base se invoca siempre con `training=False` y las capas BN se mantienen congeladas aunque se descongele su bloque. Si se dejara que BN recalculase sus estadísticas con lotes pequeños de un dataset nuevo, las activaciones que esperan las capas posteriores cambiarían bruscamente y se perdería parte de lo aprendido.

Para que la comparación entre arquitecturas sea limpia, las tres se entrenan primero con el **mismo protocolo y sin data augmentation**. El efecto de la regularización y de un *fine-tuning* más profundo se estudia después sobre la arquitectura seleccionada (apartados 4.4 y 4.5).

```python
class CaffePreprocess(layers.Layer):
    # Preprocesamiento de VGG16 y ResNet50: RGB -> BGR y resta de la media de ImageNet por canal.
    def call(self, x):
        x = tf.reverse(x, axis=[-1])
        return x - tf.constant([103.939, 116.779, 123.68], dtype=x.dtype)

BACKBONES = {
    'VGG16':       dict(fn=keras.applications.VGG16,       mode='caffe', ft_layers=['block5_conv1', 'block4_conv1']),
    'ResNet50':    dict(fn=keras.applications.ResNet50,    mode='caffe', ft_layers=['conv5_block1_1_conv', 'conv4_block1_1_conv']),
    'MobileNetV2': dict(fn=keras.applications.MobileNetV2, mode='tf',    ft_layers=['block_14_expand', 'block_7_expand']),
}

TL_EPOCHS = 2 if MODO_RAPIDO else 15
FT_EPOCHS = 2 if MODO_RAPIDO else 15
TL_PATIENCE, FT_PATIENCE = 4, 5
TL_LR, FT_LR = 1e-3, 1e-5

def build_pretrained(backbone, augment=False, dropout=0.2):
    keras.utils.set_random_seed(SEED)
    spec = BACKBONES[backbone]
    base = spec['fn'](include_top=False, weights='imagenet', input_shape=(IMG_SIZE, IMG_SIZE, 3))
    base.trainable = False                                   # transfer learning: base congelada
    inputs = keras.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
    x = inputs
    if augment:
        x = build_augmenter()(x)
    if spec['mode'] == 'caffe':
        x = CaffePreprocess()(x)
    else:
        x = layers.Rescaling(1.0 / 127.5, offset=-1.0)(x)
    x = base(x, training=False)                              # BN de la base siempre en modo inferencia
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(dropout)(x)
    outputs = layers.Dense(NUM_CLASSES, activation='softmax')(x)
    return keras.Model(inputs, outputs), base

def unfreeze_from(base, layer_name):
    # Descongela la base desde la capa indicada hasta el final (las capas BN permanecen congeladas).
    names = [l.name for l in base.layers]
    if layer_name in names:
        start = names.index(layer_name)
    else:
        start = int(0.7 * len(names))
        print('AVISO: capa %s no encontrada; se descongela el ultimo 30%% de las capas' % layer_name)
    base.trainable = True
    for i, layer in enumerate(base.layers):
        layer.trainable = (i >= start) and not isinstance(layer, layers.BatchNormalization)
    print('Capas de la base descongeladas: %d de %d (desde %s)' % (len(names) - start, len(names), names[start]))

MODELS, STATE, PHASES = {}, {}, {}      # por variante: (modelo, base), fase cuyos pesos estan cargados, fases ejecutadas

def transfer_learning(tag, backbone, augment=False, dropout=0.2):
    model, base = build_pretrained(backbone, augment=augment, dropout=dropout)
    MODELS[tag] = (model, base)
    name = tag + '_TL'
    desc = '%s: transfer learning (base congelada%s)' % (backbone, ', con augmentation' if augment else '')
    run_experiment(name, model, desc, 'pretrained', lr=TL_LR, epochs=TL_EPOCHS, patience=TL_PATIENCE)
    STATE[tag], PHASES[tag] = name, [name]

def finetune_stage(tag, from_layer, suffix, lr=FT_LR):
    # Descongela desde from_layer y continua el entrenamiento. Si la validacion empeora, se restauran los pesos previos.
    model, base = MODELS[tag]
    prev = STATE[tag]
    unfreeze_from(base, from_layer)
    snapshot = model.get_weights()
    name = tag + '_' + suffix
    run_experiment(name, model, 'fine-tuning desde %s (lr=%g), partiendo de %s' % (from_layer, lr, prev),
                   'pretrained', lr=lr, epochs=FT_EPOCHS, patience=FT_PATIENCE)
    PHASES[tag].append(name)
    if RESULTS[name]['val_acc'] < RESULTS[prev]['val_acc']:
        model.set_weights(snapshot)
        print('La fase %s no mejora la validacion de %s: se restauran los pesos anteriores.' % (name, prev))
    else:
        STATE[tag] = name
    print('Estado vigente de %s: %s (val_acc = %.1f%%)' % (tag, STATE[tag], 100 * RESULTS[STATE[tag]]['val_acc']))
```

## 4.2 Transfer learning y fine-tuning de las tres arquitecturas

Para cada arquitectura: (1) *transfer learning* con la base congelada y (2) *fine-tuning* del último bloque convolucional (`block5` en VGG16, `conv5` en ResNet50 y los tres últimos bloques residuales invertidos más la convolución final en MobileNetV2). En las gráficas, la línea punteada vertical separa las dos fases.

**Hipótesis.** Ya con la base congelada, las tres redes deberían superar al mejor modelo entrenado desde cero y hacerlo en muy pocas épocas, porque las características de ImageNet son directamente útiles para fotografías de flores. El *fine-tuning* debería aportar una mejora adicional más modesta.

```python
transfer_learning('VGG16', 'VGG16')
finetune_stage('VGG16', BACKBONES['VGG16']['ft_layers'][0], 'FT1')
plot_history(PHASES['VGG16'], 'VGG16: transfer learning + fine-tuning del ultimo bloque')
```

```python
transfer_learning('ResNet50', 'ResNet50')
finetune_stage('ResNet50', BACKBONES['ResNet50']['ft_layers'][0], 'FT1')
plot_history(PHASES['ResNet50'], 'ResNet50: transfer learning + fine-tuning del ultimo bloque')
```

* **Recorrido de la estrategia 2:** el *transfer learning* aporta la mayor parte del rendimiento (91,5 %), el *fine-tuning* añade alrededor de 1 punto, el *augmentation* reduce el sobreajuste a la mitad y el *fine-tuning* progresivo aporta una mejora marginal.

# 5. Evaluación sobre test y comparación global

Hasta este punto todas las decisiones se han tomado con el conjunto de validación. Ahora, **con los modelos ya fijados**, se usa por primera y única vez el conjunto de test, que estima el rendimiento sobre datos nunca vistos.

Para cada modelo se calcula: *accuracy* con su **intervalo de confianza del 95 %** (método de Wilson; con un test de unos cientos de imágenes la incertidumbre no es despreciable), F1 macro, informe por clase (*precision*, *recall*, F1) y matriz de confusión.

```python
TEST = {}

def wilson_interval(p, n, z=1.96):
    denom = 1 + z ** 2 / n
    centre = (p + z ** 2 / (2 * n)) / denom
    half = z * np.sqrt(p * (1 - p) / n + z ** 2 / (4 * n ** 2)) / denom
    return centre - half, centre + half

def evaluate_on_test(model, name):
    probs = model.predict(test_ds, verbose=0)
    y_pred = probs.argmax(axis=1)
    acc = float((y_pred == y_test).mean())
    lo, hi = wilson_interval(acc, len(y_test))
    TEST[name] = dict(acc=acc, lo=lo, hi=hi, macro_f1=float(f1_score(y_test, y_pred, average='macro')),
                      f1_clase=f1_score(y_test, y_pred, average=None), y_pred=y_pred, probs=probs)
    print('=' * 80)
    print('TEST | %s | accuracy = %.1f%%  (IC 95%%: %.1f%% - %.1f%%) | F1 macro = %.3f' % (
        name, 100 * acc, 100 * lo, 100 * hi, TEST[name]['macro_f1']))
    print(classification_report(y_test, y_pred, target_names=CLASS_NAMES, digits=3))
    cm = confusion_matrix(y_test, y_pred)
    fig, ax = plt.subplots(figsize=(5, 4.2))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False, xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, ax=ax)
    ax.set_xlabel('clase predicha'); ax.set_ylabel('clase real'); ax.set_title('Matriz de confusion (test) - ' + name)
    plt.tight_layout(); plt.show()

def show_errors(name, n=8):
    # Errores de test cometidos con mayor confianza.
    y_pred, probs = TEST[name]['y_pred'], TEST[name]['probs']
    wrong = np.where(y_pred != y_test)[0]
    wrong = wrong[np.argsort(-probs[wrong].max(axis=1))][:n]
    if len(wrong) == 0:
        print('Sin errores en test.'); return
    fig, axes = plt.subplots(1, len(wrong), figsize=(2.3 * len(wrong), 2.9), squeeze=False)
    for ax, i in zip(axes[0], wrong):
        img, _ = load_image(test_df.loc[i, 'path'], 0)
        ax.imshow(img.numpy()); ax.axis('off')
        ax.set_title('real: %s\npred: %s (%.0f%%)' % (CLASS_NAMES[y_test[i]], CLASS_NAMES[y_pred[i]],
                                                    100 * probs[i].max()), fontsize=8)
    plt.suptitle('Errores con mayor confianza - ' + name, y=1.08); plt.show()
```

```python
evaluate_on_test(scratch_model, 'Scratch (S_final)')
show_errors('Scratch (S_final)')
```

Además de comparar las dos *accuracies*, se aplica la **prueba de McNemar** (versión exacta) sobre las predicciones de test. Es la prueba adecuada para comparar dos clasificadores evaluados sobre las mismas imágenes: solo tiene en cuenta los casos en los que uno acierta y el otro falla, y contrasta si esa asimetría puede deberse al azar.

```python
S_NAME = 'Scratch (S_final)'
s_res, p_res = RESULTS['S_final'], RESULTS[FINAL_PRETRAINED]
scratch_time = sum(RESULTS[n]['tiempo_s'] for n in RESULTS if RESULTS[n]['estrategia'] == 'scratch')
pre_time_final = sum(RESULTS[p]['tiempo_s'] for p in PHASES[BEST_TAG])

comparison = pd.DataFrame({
    'From scratch': {
        'modelo': 'CNN propia (S_final)',
        'test_acc_%': round(100 * TEST[S_NAME]['acc'], 1),
        'IC95_%': '%.1f - %.1f' % (100 * TEST[S_NAME]['lo'], 100 * TEST[S_NAME]['hi']),
        'test_F1_macro': round(TEST[S_NAME]['macro_f1'], 3),
        'val_acc_%': round(100 * s_res['val_acc'], 1),
        'train_acc_%': round(100 * s_res['train_acc'], 1),
        'gap_train_val_pts': round(100 * s_res['gap'], 1),
        'params_totales': s_res['params'],
        'epocas_modelo_final': s_res['epocas'],
        'tiempo_modelo_final_s': round(s_res['tiempo_s']),
        'tiempo_toda_la_experimentacion_s': round(scratch_time)},
    'Preentrenado': {
        'modelo': '%s (%s)' % (BEST_ARCH, FINAL_PRETRAINED),
        'test_acc_%': round(100 * TEST[PRE_NAME]['acc'], 1),
        'IC95_%': '%.1f - %.1f' % (100 * TEST[PRE_NAME]['lo'], 100 * TEST[PRE_NAME]['hi']),
        'test_F1_macro': round(TEST[PRE_NAME]['macro_f1'], 3),
        'val_acc_%': round(100 * p_res['val_acc'], 1),
        'train_acc_%': round(100 * p_res['train_acc'], 1),
        'gap_train_val_pts': round(100 * p_res['gap'], 1),
        'params_totales': p_res['params'],
        'epocas_modelo_final': sum(RESULTS[p]['epocas'] for p in PHASES[BEST_TAG]),
        'tiempo_modelo_final_s': round(pre_time_final),
        'tiempo_toda_la_experimentacion_s': round(sum(RESULTS[n]['tiempo_s'] for n in RESULTS if RESULTS[n]['estrategia'] == 'pretrained'))}})
display(comparison)

# Prueba de McNemar exacta sobre las predicciones de test
ok_s = TEST[S_NAME]['y_pred'] == y_test
ok_p = TEST[PRE_NAME]['y_pred'] == y_test
only_s, only_p = int((ok_s & ~ok_p).sum()), int((~ok_s & ok_p).sum())
print('Imagenes de test que acierta solo la CNN from scratch  :', only_s)
print('Imagenes de test que acierta solo el modelo preentrenado:', only_p)
print('Imagenes que fallan ambos:', int((~ok_s & ~ok_p).sum()), '| que aciertan ambos:', int((ok_s & ok_p).sum()))
if only_s + only_p > 0:
    pval = stats.binomtest(min(only_s, only_p), only_s + only_p, 0.5).pvalue
    print('McNemar exacto: p-valor = %.2g' % pval)
```
